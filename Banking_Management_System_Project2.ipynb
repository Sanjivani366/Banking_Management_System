```python
# ==========================================
# 🏦 BANKEASE BANK
# Banking Management System
# ==========================================

accounts = []
transactions = []


# ==========================================
# CREATE ACCOUNT
# ==========================================

def create_account():

    print("\n===== CREATE ACCOUNT =====")

    account_number = input("Enter Account Number: ")

    # Check duplicate account number
    for account in accounts:
        if account["account_number"] == account_number:
            print("\nAccount Number already exists!")
            return

    name = input("Enter Customer Name: ")
    age = int(input("Enter Age: "))
    mobile = input("Enter Mobile Number: ")
    account_type = input("Enter Account Type (Saving/Current): ")
    initial_deposit = float(input("Enter Initial Deposit: "))

    account = {
        "account_number": account_number,
        "name": name,
        "age": age,
        "mobile": mobile,
        "account_type": account_type,
        "balance": initial_deposit
    }

    accounts.append(account)

    print("\nAccount created successfully!")


# ==========================================
# VIEW ACCOUNT
# ==========================================

def view_account():

    print("\n===== VIEW ACCOUNT =====")

    account_number = input("Enter Account Number: ")

    for account in accounts:

        if account["account_number"] == account_number:

            print("\nAccount Details")
            print("Account Number:", account["account_number"])
            print("Name:", account["name"])
            print("Age:", account["age"])
            print("Mobile:", account["mobile"])
            print("Account Type:", account["account_type"])
            print("Balance:", account["balance"])

            return

    print("Account not found!")


# ==========================================
# DEPOSIT MONEY
# ==========================================

def deposit_money():

    print("\n===== DEPOSIT MONEY =====")

    account_number = input("Enter Account Number: ")

    for account in accounts:

        if account["account_number"] == account_number:

            amount = float(input("Enter Deposit Amount: "))

            if amount <= 0:
                print("Amount must be greater than 0!")
                return

            account["balance"] += amount

            transactions.append({
                "type": "Deposit",
                "account_number": account_number,
                "amount": amount
            })

            print("\nMoney deposited successfully!")
            print("New Balance:", account["balance"])

            return

    print("Account not found!")


# ==========================================
# WITHDRAW MONEY
# ==========================================

def withdraw_money():

    print("\n===== WITHDRAW MONEY =====")

    account_number = input("Enter Account Number: ")

    for account in accounts:

        if account["account_number"] == account_number:

            amount = float(input("Enter Withdrawal Amount: "))

            if amount <= 0:
                print("Amount must be greater than 0!")
                return

            if amount <= account["balance"]:

                account["balance"] -= amount

                transactions.append({
                    "type": "Withdraw",
                    "account_number": account_number,
                    "amount": amount
                })

                print("\nMoney Withdrawn Successfully!")
                print("New Balance:", account["balance"])

            else:
                print("Insufficient Balance!")

            return

    print("Account not found!")


# ==========================================
# CHECK BALANCE
# ==========================================

def check_balance():

    print("\n===== CHECK BALANCE =====")

    account_number = input("Enter Account Number: ")

    for account in accounts:

        if account["account_number"] == account_number:

            print("\nAccount Number:", account["account_number"])
            print("Account Holder:", account["name"])
            print("Current Balance:", account["balance"])

            return

    print("Account not found!")


# ==========================================
# TRANSFER MONEY
# ==========================================

def transfer_money():

    print("\n===== TRANSFER MONEY =====")

    sender_number = input("Enter Your Account Number: ")
    receiver_number = input("Enter Receiver Account Number: ")

    sender = None
    receiver = None

    # Find sender and receiver
    for account in accounts:

        if account["account_number"] == sender_number:
            sender = account

        if account["account_number"] == receiver_number:
            receiver = account

    if sender is None:
        print("Sender Account not found!")
        return

    if receiver is None:
        print("Receiver Account not found!")
        return

    amount = float(input("Enter Transfer Amount: "))

    if amount <= 0:
        print("Amount must be greater than 0!")
        return

    if amount <= sender["balance"]:

        sender["balance"] -= amount
        receiver["balance"] += amount

        transactions.append({
            "type": "Transfer",
            "from": sender_number,
            "to": receiver_number,
            "amount": amount
        })

        print("\nMoney transferred successfully!")
        print("Your New Balance:", sender["balance"])

    else:
        print("Insufficient Balance!")


# ==========================================
# TRANSACTION HISTORY
# ==========================================

def transaction_history():

    print("\n===== TRANSACTION HISTORY =====")

    if len(transactions) == 0:
        print("No transactions found!")
        return

    for transaction in transactions:

        print("\nTransaction Type:", transaction["type"])

        if transaction["type"] == "Deposit":

            print("Account Number:", transaction["account_number"])
            print("Amount:", transaction["amount"])

        elif transaction["type"] == "Withdraw":

            print("Account Number:", transaction["account_number"])
            print("Amount:", transaction["amount"])

        elif transaction["type"] == "Transfer":

            print("From Account:", transaction["from"])
            print("To Account:", transaction["to"])
            print("Amount:", transaction["amount"])


# ==========================================
# DELETE ACCOUNT
# ==========================================

def delete_account():

    print("\n===== DELETE ACCOUNT =====")

    account_number = input("Enter Account Number: ")

    for account in accounts:

        if account["account_number"] == account_number:

            accounts.remove(account)

            print("\nAccount deleted successfully!")

            return

    print("Account not found!")


# ==========================================
# MAIN MENU
# ==========================================

while True:

    print("\n=============================")
    print("     🏦 BANKEASE BANK")
    print("  Banking Management System")
    print("=============================")

    print("\n===== MAIN MENU =====")
    print("1. Create Account")
    print("2. View Account")
    print("3. Deposit Money")
    print("4. Withdraw Money")
    print("5. Check Balance")
    print("6. Transfer Money")
    print("7. Transaction History")
    print("8. Delete Account")
    print("9. Exit")

    choice = input("\nEnter Your Choice: ")

    if choice == "1":
        create_account()

    elif choice == "2":
        view_account()

    elif choice == "3":
        deposit_money()

    elif choice == "4":
        withdraw_money()

    elif choice == "5":
        check_balance()

    elif choice == "6":
        transfer_money()

    elif choice == "7":
        transaction_history()

    elif choice == "8":
        delete_account()

    elif choice == "9":
        print("\nThank you for using BANKEASE BANK! 🏦")
        break

    else:
        print("\nInvalid Choice! Please try again.")
```
